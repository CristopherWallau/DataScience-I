```python
import pandas as pd
import altair as alt
from vega_datasets import data
import os
import kagglehub
alt.data_transformers.disable_max_rows()
```

```python
# Download latest version
path = kagglehub.dataset_download("suhanigupta04/employee-mental-health-and-burnout-dataset")

print("Path to dataset files:", path)
```

```python
arquivos = os.listdir(path)
print("Arquivos disponíveis no dataset:",arquivos)

caminho_csv = os.path.join(path, "tech_mental_health_burnout.csv")
df = pd.read_csv(caminho_csv)
df.head()
```

```python
df.describe()
```

```python
df.info()
```

```python
# Filter data for burnout score >= 6
filtered_df = df[df['stress_level'] >= 6]

# Group by Work_Mode and count
burnout_by_mode = filtered_df.groupby('work_mode').size().reset_index(name='Count')

# Create bar chart with Altair
chart = alt.Chart(burnout_by_mode).mark_bar().encode(
    x='work_mode:N',
    y='Count:Q',
    tooltip=['work_mode', 'Count']
).properties(
    title='Stress Level >= 6 by Work Mode'
)

chart.display()
```

```python
work_mode_counts = df['work_mode'].value_counts().reset_index(name='Count').rename(columns={'index': 'work_mode'})

alt.Chart(work_mode_counts).mark_bar().encode(
    x='work_mode:N',
    y='Count:Q',
    tooltip=['work_mode', 'Count']
).properties(
    title='Number of Employees per Work Mode'
).display()
```

```python
# Filter data for stress level >= 6
filtered_df = df[df['stress_level'] >= 6]

# Group by Company_Size and count
burnout_by_size = filtered_df.groupby('company_size').size().reset_index(name='Count')

# Create bar chart with Altair
chart = alt.Chart(burnout_by_size).mark_bar().encode(
    x='company_size:N',
    y='Count:Q',
    tooltip=['company_size', 'Count']
).properties(
    title='Stress Level >= 6 by Company Size'
)

chart.display()
```
